
#Partition 

1. it is optimization Techinque in databricks that is used to divide large dataset into smaller files based on columns.
2. It will improve performance.


###Read Hotel data

```python

hotel_df = spark.read.format('csv')\
    .option("header", True)\
    .option("inferSchema", True)\
    .load("/Volumes/hotal_catalog/hotel_schema/hotel_datafiles/hotels.csv")
display(hotel_df)
```

```python
hotel_df.columns
```


###Cleaning DataFrame Column

```python
import re

for old_col in hotel_df.columns:
    new_col = re.sub(r"[^a-zA-Z0-9]+", "_", old_col)
    new_col = new_col.strip("_").lower()
    hotel_df = hotel_df.withColumnRenamed(old_col, new_col)

hotel_df.printSchema()
```

```python
display(hotel_df)
```


###Partition in Azure Databrics based on Country_name

```python
hotel_df.write.format("delta").mode("overwrite")\
.partitionBy("country_name")\
.save("/Volumes/hotal_catalog/hotel_schema/hotel_datafiles/Hotal_partitionBy_country")
```
