```python

#import all module
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
#Create dataframe for employee csv data
employee_df = spark.read.format("csv")\
    .option('header','true')\
        .option('inferSchema','true')\
            .load('/Volumes/employee_catalog/employee_schema/employee_datafiles/employee.csv')\
                .withColumn('end_date',(to_date('end_date','yyyy/MM/dd')))

display(employee_df)
```

```python
#Create Data Frame for employee2 csv data with Changing data
employee_df2 = spark.read.format('csv')\
    .option('header','true')\
        .option('inferSchema','true')\
            .load('/Volumes/employee_catalog/employee_schema/employee_datafiles/employee2.csv')

display(employee_df2)
```

```python

#Added History Column for employee2
employee_df2A = employee_df2.join(employee_df,'employee_id','left_semi')\
    .withColumn('Start_date',current_date())\
    .withColumn('end_date',lit(None).cast('Date'))\
    .withColumn('IsActive',lit(1))
display(employee_df2A)
```

```python
#Added end_date and ISActive column for History data
employee_df3 = employee_df.join(employee_df2,'employee_id','left_semi')\
    .withColumn('end_date',current_date() - 1)\
    .withColumn('IsActive',lit(0))
display(employee_df3)
```

```python
#Pull Other Than matching data with History employee Data
employee_df4 = employee_df2.join(employee_df,'employee_id','left_anti')\
    .withColumn('joining_date',current_date())\
    .withColumn('Start_date',current_date())\
    .withColumn('end_date',lit(None).cast('Date'))\
    .withColumn('IsActive',lit(1))
display(employee_df4)
```

```python
#Pull other than Changing data
employee_df5 = employee_df.join(employee_df2,'employee_id','left_anti')
display(employee_df5)
```

```python
from functools import reduce
from pyspark.sql.functions import DataFrame
```

```python
def unionall(*a):
    return reduce(DataFrame.union,a)
df_final=unionall(employee_df2A,employee_df3,employee_df4,employee_df5)
display(df_final.orderBy("employee_id"))
```

```python
#overwrite data in employee_history delta table
df_final.write.format('csv').mode('overwrite').save('/Volumes/employee_catalog/employee_schema/employee_datafiles/employee_data')
```
