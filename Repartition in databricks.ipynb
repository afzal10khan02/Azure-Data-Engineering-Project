#Repartition

1. Repartition in Databricks is used to increase or decrease no of partition in Dataframe.
2. It's perform full shuffling of data to distribute it to same partition.

**Note** - A full shuffle in repartition() means that Spark redistributes data across all executors to create new partitions. During this process, rows are moved over the network so they can be evenly distributed or grouped by a specified column.

####When to use repartition()

1. Before writing data to control the number of output files.
2. Before joins or aggregations to improve data distribution.
3. When you need to increase the number of partitions for better parallelism.

###Create Dataframe for Hotel_dataset

```python
Hotel_repartition_df = spark.read.format("csv")\
    .option("header","true")\
    .option("inferSchema","true")\
        .load("/Volumes/hotal_catalog/hotel_schema/hotel_datafiles/hotels.csv")

display(Hotel_repartition_df)
```

###Cleaning DataFrame Column

```python
import re

for old_Df in Hotel_repartition_df.columns:
    new_Df = re.sub(r"[^a-zA-Z0-9]+", "_", old_Df)
    new_Df = new_Df.strip("_").lower()
    Hotel_repartition_df = Hotel_repartition_df.withColumnRenamed(old_Df,new_Df)

Hotel_repartition_df.printSchema()
```

###Display Dataframe

```python
display(Hotel_repartition_df)
```

####Write data on Volume in Delta format

```python
Hotel_repartition_df.repartition(5).write.format("delta").mode("overwrite").save("/Volumes/hotal_catalog/hotel_schema/hotel_datafiles/Hotel_repartition")
```
